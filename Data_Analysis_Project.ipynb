```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Download a sample dataset (E-commerce trends)
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

# 2. Display basic summary statistics
print("--- Dataset Head ---")
print(df.head())

print("\n--- Summary Statistics ---")
print(df.describe())

# 3. Create a Visualization
plt.figure(figsize=(8, 5))
plt.scatter(df['total_bill'], df['tip'], alpha=0.7, color='teal')
plt.title('Total Bill vs Tip Amount')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.grid(True)
plt.savefig('bill_vs_tip.png')
plt.show()
```

# Key Insights from E-Commerce Data
- **Strong Positive Correlation:** Higher total bill amounts directly lead to higher tip amounts.
- **Outliers:** A few customers spent over $40 but tipped less than $3.
